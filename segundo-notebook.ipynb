```python
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv(
    "vendas_eletronicos.csv"
)
```

```python
grafico=sns.histplot(data=df,x="categoria",kde=True,bins=4)
grafico.set_title("categoria")
grafico.set_xlabel("preços")
grafico.set_ylabel("Quantidade")

```

```python
grafico=sns.boxplot(data=df,x="categoria",y="preco_unitario")
```
